```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('train.txt', sep=';', header=None, names=['text', 'emotion'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['emotion'].unique()
```

```python
unique_emotions=df['emotion'].unique()
emotion_number ={}
i =0
for emotion in unique_emotions:
    emotion_number[emotion]=i
    i+=1

df['emotion'] = df['emotion'].map(emotion_number)
```

```python
df
```

CONVERTING TEXT INTO LOWERCASE

```python
df
```

```python
df['text']=df['text'].apply(lambda  x : x.lower())
```

REMOVING PUNCTUATION

```python
import string
```

```python
def removing_punch(text):
  return text.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(removing_punch)
```

REMOVING NUMBERS

```python
def remove_number(text):
  new=''
  for i in text:
    if not i.isdigit():
        new=new+i
  return new

df['text']=df['text'].apply(remove_number)
```

REMOVE URL AND LINK

```python
import re
def remove_urls(text):
  url_pattern = re.compile(r'https?://\S+|www\.\S+')
  return url_pattern.sub(r'', text)
df['text'] = df['text'].apply(remove_urls)
```

REMOVE HTML TAGS

```python

def remove_html_tags(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
df['text'] = df['text'].apply(remove_html_tags)
```

REMOVE EMOJIES AND CHARASCTRER

```python
def remove_emojis(text):
  new=''
  for i in text:
    if i.isascii(): # This condition means: if the character IS ASCII, then add it to 'new'.
        new=new+i
  return new

df['text']=df['text'].apply(remove_emojis)
```

```python
df.head()
```

REMOVE STOPWORDS

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
df.head()
```

```python
len(stop_words)
```

```python

def remove(txt):
  words = txt.split()
  cleaned = []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)

  return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)

nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)

pred_bow = nb_model.predict(X_test_bow)
print(accuracy_score(y_test,pred_bow))
```

```python
pred_bow
```

```python
tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

nb2_model = MultinomialNB()
nb2_model.fit(X_train_tfidf, y_train)

pred_tfidf = nb2_model.predict(X_test_tfidf)
print(accuracy_score(y_test,pred_tfidf))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
LogisticRegression_model = LogisticRegression()
LogisticRegression_model.fit(X_train_tfidf, y_train)
```

```python
log_pred = LogisticRegression_model.predict(X_test_tfidf)
```
